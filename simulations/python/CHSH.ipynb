# Протокол CHSH — проверка неравенств Белла

Протокол Клаузера — Хорна — Шимони — Хольта (Clauser — Horne — Shimony — Holt, 1969) —
стандартная экспериментальная схема проверки неравенств Белла на паре запутанных частиц.

Каждая из двух сторон выбирает одну из двух настроек анализатора:

- сторона A: углы $a_1 = 0^\circ$, $a_2 = 90^\circ$;
- сторона B: углы $b_1 = 45^\circ$, $b_2 = 135^\circ$.

Комбинация корреляций

$$S = E(a_1,b_1) - E(a_1,b_2) + E(a_2,b_1) + E(a_2,b_2)$$

для любой локальной теории со скрытыми параметрами ограничена величиной $|S| \le 2$,
тогда как квантовая механика для синглетного состояния даёт $|S| = 2\sqrt{2} \approx 2.828$
(граница Цирельсона) при корреляторе $E(\theta) = -\cos 2\theta$.

В настоящем ноутбуке проверяется, какие значения $S$ даёт корреляционная функция,
порождаемая моделью упругого вакуума (см. `theory/09-nelokalnost.md` и
`theory/09-1-epr-akusticheskiy-kabel.md`), и сопоставляется с квантовым пределом.

Связанный скрипт: `bell.py`.

```python
import numpy as np

angles_1 = [0, 90]
angles_2 = [45, 135]

def sn(x):
    y = np.cos(np.radians(x))
    return y*y*np.sign(y)

def prob(y1, y2):
    k=np.sqrt((np.abs(y1)+np.abs(y2))/2)
    return k*(y1+y2)/2       

def f(x, dx):
    x = x + dx
    y1 = sn(x)
    y2 = sn(2*x + dx)
    return prob(y1, y2)
    
print("Для функции cos(4*theta):")
for a1 in angles_1:
    for a2 in angles_2:
        diff = a1 - a2
        dx = 1.125
        corr = f(a1/2, dx)*f(a2/2, dx)
        print(f"Пара ({a1}°, {a2}°): Разница = {diff}°, Корреляция = {corr:.2f}")
```
