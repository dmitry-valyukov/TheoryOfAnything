```python
import numpy as np

angles_1 = [0, 90]
angles_2 = [45, 135]

def sn(x):
    y = np.cos(np.radians(x))
    return y*y*np.sign(y)

def prob(y1, y2):
    k=np.sqrt((np.abs(y1)+np.abs(y2))/2)
    return k*(y1+y2)/2       

def f(x, dx):
    x = x + dx
    y1 = sn(x)
    y2 = sn(2*x + dx)
    return prob(y1, y2)
    
print("Для функции cos(4*theta):")
for a1 in angles_1:
    for a2 in angles_2:
        diff = a1 - a2
        dx = 1.125
        corr = f(a1/2, dx)*f(a2/2, dx)
        print(f"Пара ({a1}°, {a2}°): Разница = {diff}°, Корреляция = {corr:.2f}")
```
